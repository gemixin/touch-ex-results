# Generating CSV Tables

This notebook creates CSV tables from experiment results.

```python
from pathlib import Path
import pandas as pd
```

### Accuracy and F1 Tables for Classification
Run the cell below to generate compact results tables for the classification experiments. Each table contains accuracy and weighted F1 scores for the test, test unseen matched and test unseen related splits.

```python
RESULT_FOLDER_NAMES = ("object_classify", "object_region_classify", "force_level_classify")
EXPORT_DIR = Path("../acc_f1_tables")
TABLE_COLUMNS = [
    "run_name",
    "test_acc",
    "test_weighted_f1_avg",
    "test_unseen_matched_acc",
    "test_unseen_matched_weighted_f1_avg",
    "test_unseen_related_acc",
    "test_unseen_related_weighted_f1_avg",
]
# Export one CSV file per experiment
for result_folder in RESULT_FOLDER_NAMES:
    # Build the path to the combined experiment results for this result set
    results_file = Path("../results") / result_folder / "experiments.parquet"

    # Create a separate export folder for this result set
    result_export_dir = EXPORT_DIR / result_folder
    result_export_dir.mkdir(parents=True, exist_ok=True)

    # Each group contains every model run belonging to one named experiment
    results_df = pd.read_parquet(results_file)
    for (experiment_number, experiment_name), experiment_df in results_df.groupby(
        ["experiment_number", "experiment_name"], sort=False
    ):
        table = (
            experiment_df[TABLE_COLUMNS]
            .sort_values("test_acc", ascending=False)
            .reset_index(drop=True)
        )
        # Pad the experiment number so files remain in numerical order
        output_file = result_export_dir / f"{experiment_number:03d}_{experiment_name}.csv"
        table.to_csv(output_file, index=False)
        print(f"Saved {output_file}")
```

### MAE, RMSE, and R² Tables for Regression
Run the cell below to generate compact results tables for the regression experiments. Each table contains MAE, RMSE, and R² scores for the test, test unseen matched and test unseen related splits.

```python
REGRESSION_RESULT_FOLDER_NAMES = ("force_n_regress",)
REGRESSION_EXPORT_DIR = Path("../mae_rmse_r2_tables")
REGRESSION_TABLE_COLUMNS = [
    "model_type",
    "test_mae",
    "test_rmse",
    "test_r2",
    "test_unseen_matched_mae",
    "test_unseen_matched_rmse",
    "test_unseen_matched_r2",
    "test_unseen_related_mae",
    "test_unseen_related_rmse",
    "test_unseen_related_r2",
]
# Export one CSV file per experiment
for result_folder in REGRESSION_RESULT_FOLDER_NAMES:
    # Build the path to the combined experiment results for this result set
    results_file = Path("../results") / result_folder / "experiments.parquet"

    # Create a separate export folder for this result set
    result_export_dir = REGRESSION_EXPORT_DIR / result_folder
    result_export_dir.mkdir(parents=True, exist_ok=True)

    # Each group contains every model run belonging to one named experiment
    results_df = pd.read_parquet(results_file)
    for (experiment_number, experiment_name), experiment_df in results_df.groupby(
        ["experiment_number", "experiment_name"], sort=False
    ):
        table = (
            experiment_df[REGRESSION_TABLE_COLUMNS]
            .sort_values("test_mae")
            .reset_index(drop=True)
        )
        # Pad the experiment number so files remain in numerical order
        output_file = result_export_dir / f"{experiment_number:03d}_{experiment_name}.csv"
        table.to_csv(output_file, index=False)
        print(f"Saved {output_file}")
```
