## View Experiment Results

This notebook allows you to view an experiments Parquet file as a DataFrame.


```python
# Imports
import pandas as pd
from IPython.display import display
```

```python
# Choose an experiments file; the path is relative to this notebook's directory
PARQUET_PATH = "../results/final_evaluation/object_classification/experiments.parquet"
```

```python
experiments_df = pd.read_parquet(PARQUET_PATH)
# Show all metadata columns without changing global pandas display settings
with pd.option_context("display.max_columns", None):
    display(experiments_df)
```
